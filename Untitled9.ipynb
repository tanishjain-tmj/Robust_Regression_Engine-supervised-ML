#part A: THEORY:

Q1. What is Regularization in Machine Learning? Why is it needed?

Ans. Regularization is a technique used to prevent overfitting by adding a penalty to the model's complexity. It helps improve generalization and prediction accuracy on unseen data.

Q2. Difference between Ridge Regression (L2) and Lasso Regression (L1).

Ans. Ridge (L2)	Lasso (L1)
Uses squared coefficients as a penalty,Shrinks coefficients towards zero,Keeps all features

Lasso (L1) Uses absolute coefficients as a penalty, Can make coefficients exactly zero, Performs feature selection


Q3. What is Cross-Validation and why is it important?

Ans. Cross-Validation is a technique used to evaluate model performance by dividing data into training and validation sets multiple times. It helps measure model stability and reduces the risk of overfitting.

Q4. Explain the following Cross-Validation techniques.

Ans.
a) K-Fold Cross-Validation:
Divides the dataset into K equal folds. Each fold is used once for validation while the remaining folds are used for training.

b) Stratified K-Fold Cross-Validation:
Divides data into folds while maintaining a similar distribution of target categories or ranges in each fold.

c) Leave-One-Out Cross-Validation (LOOCV):
Uses one observation for validation and all remaining observations for training. This process repeats for every observation.

d) Time Series Split:
Divides data according to chronological order. Earlier observations are used for training and later observations for validation.


Q5. Why are Tree-Based Models less sensitive to Feature Scaling?

Ans.Tree-based models make decisions using feature-based conditions and thresholds rather than distances or magnitudes. Therefore, scaling features generally does not affect their splitting decisions.

```python
#features and target variable :
x= df.drop(['property_id','sale_date','house_price_inr'],axis=1)
y= df['house_price_inr']
print("features:",x.columns.tolist())
print("Traget variable:",y.name)
print("features shape:",x.shape)
print("target variable shape",y.shape)
```

```python
#train-test split
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
print("training data size:",x_train.shape,"\ntesting data size:",x_test.shape)
```

```python
#scalling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled= scaler.fit_transform(x_train)
x_test_scaled= scaler.transform(x_test)
```

#Part C:Regularized linear Model:

```python
#Ridge Regression L2:
from sklearn.linear_model import Ridge
ridge_model= Ridge(alpha=1.0)
ridge_model.fit(x_train_scaled,y_train)
y_pred_ridge= ridge_model.predict(x_test_scaled)
```

```python
#lasso regresion L1:
from sklearn.linear_model import Lasso
lasso_model= Lasso(alpha=1.0,max_iter=1000)
lasso_model.fit(x_train_scaled,y_train)
y_pred_lasso= lasso_model.predict(x_test_scaled)
```

```python
#Tune the regularization parameter using cross validation:
from sklearn.model_selection import GridSearchCV
```

```python
#tune ridge regression:
ridge_params={'alpha':[0.01,0.1,1,10,100]}
ridge_grid=GridSearchCV(Ridge(),ridge_params,cv=5,scoring='r2')
ridge_grid.fit(x_train_scaled,y_train)
print("Best Ridge alpha:",ridge_grid.best_params_)
print("best Ridge CV Score:",ridge_grid.best_score_)
```

```python
#tune lasso regression:
lasso_params={'alpha':[0.01,0.1,1,10,100]}
lasso_grid=GridSearchCV(Lasso(max_iter=10000),lasso_params,cv=5,scoring='r2')
lasso_grid.fit(x_train_scaled,y_train)
print("Best lasso alpha:",lasso_grid.best_params_)
print("best lasso CV Score:",lasso_grid.best_score_)
```

```python
#comparing ridge and lasso:
# training error:
from sklearn.metrics import mean_squared_error
best_ridge= ridge_grid.best_estimator_
best_lasso= lasso_grid.best_estimator_

ridge_train_pred= best_ridge.predict(x_train_scaled)
lasso_train_pred= best_lasso.predict(x_train_scaled)

ridge_train_error= mean_squared_error(y_train,ridge_train_pred)
print("Ridge training error:",ridge_train_error)
lasso_train_error= mean_squared_error(y_train,lasso_train_pred)
print("Lasso training error:",lasso_train_error)
```

```python
#validation error:
from sklearn.model_selection import cross_val_score
ridge_val_error =-cross_val_score(best_ridge,x_train_scaled,y_train,cv=5,scoring='neg_mean_squared_error').mean()
print("Ridge validation Error:",ridge_val_error)

lasso_val_error =-cross_val_score(best_lasso,x_train_scaled,y_train,cv=5,scoring='neg_mean_squared_error').mean()
print("Lasso validation Error:",lasso_val_error)
```

```python
#feature coefficient behaviour:
coefficient_comparison = pd.DataFrame({'feature': x.columns,
                                       'Ridge coefficients': best_ridge.coef_,
                                       'Lasso coefficients': best_lasso.coef_})
print(coefficient_comparison)
```

#part D: Cross validation strategies

```python
#kfold cross validation:
from sklearn.model_selection import KFold
kfold= KFold(n_splits=5,shuffle= True, random_state=42)
kfold_scores= cross_val_score(best_ridge, x_train_scaled,y_train,cv= kfold,scoring='neg_mean_squared_error')#here best_ridge is used bcz its the best_estimator
kfold_mse= -kfold_scores.mean()
print("k-fold cross-validation")
print("Individual Fold Scores:",-kfold_scores)
print("Average MSE:",kfold_mse)
```

```python
#stratified kfold cv:
from sklearn.model_selection import StratifiedKFold
y_bins= pd.qcut(y_train,q=5,labels=False)#dividing house prices into 5 bins
stratified_kfold= StratifiedKFold(n_splits=5,shuffle= True,random_state= 42)
stratified_scores= cross_val_score(best_ridge,x_train_scaled,y_train,cv=stratified_kfold.split(x_train_scaled,y_bins),scoring='neg_mean_squared_error')
stratified_mse= -stratified_scores.mean()
print("Stratified K-fold scores:",-stratified_scores)
print("Average Stratified Mse:", stratified_mse)
```

```python
#leave one out cross validation:
from sklearn.model_selection import LeaveOneOut
loo= LeaveOneOut()
loo_scores= cross_val_score(best_ridge, x_train_scaled, y_train, cv=loo, scoring='neg_mean_squared_error')
loo_mse= -loo_scores.mean()
print("LOOCV average MSE:",loo_mse)
print("total number of folds:", len(loo_scores))
```

```python
#time series split:
from sklearn.model_selection import TimeSeriesSplit
from sklearn.pipeline import make_pipeline
#sort training data according to sale_date
train_dates= df.loc[x_train.index,'sale_date']
train_order= train_dates.sort_values().index
x_train_time= x_train.loc[train_order]
y_train_time= y_train.loc[train_order]

tscv= TimeSeriesSplit(n_splits=5)
#creating a pipeline to scale and trainridge regression
time_model= make_pipeline(StandardScaler(),Ridge(alpha=best_ridge.alpha))
#applying time series cv
time_scores= cross_val_score(time_model,x_train_time,y_train_time,cv= tscv, scoring= 'neg_mean_squared_error')
time_mse= -time_scores.mean()
print("Time Series Split Scores:", -time_scores)
print("Average Time series mSE:",time_mse)
```

```python
#comparing MSE of all four cv techniques
cv_comparison = pd.DataFrame({'cross validation techniques':['k-fold','stratified k-fold','LOOCV','Time Series split'],
                              'mean squared error':[kfold_mse,stratified_mse,loo_mse,time_mse]})
print(cv_comparison)
```

Model performance was evaluated using MSE across all four cross-validation techniques. Stratified K-Fold achieved the lowest MSE (6.2997 × 10¹²), followed closely by LOOCV (6.2984 × 10¹²), indicating slightly better prediction performance. K-Fold produced a marginally higher error, while Time Series Split recorded the highest MSE (6.3181 × 10¹²), possibly due to its chronological splitting and smaller training sets in earlier folds. Overall, the results are quite close, suggesting that the Ridge Regression model maintains relatively consistent performance across different validation strategies

#part E : Tree Based Regression model

```python
#decision tree regression:
from sklearn.tree import DecisionTreeRegressor
dt= DecisionTreeRegressor(max_depth=6,random_state=42)
dt.fit(x_train,y_train)
y_pred_dt= dt.predict(x_test)

compare = pd.DataFrame({'actual prices':y_test.head().values,
                        'predicted prices':y_pred_dt[:5]})
print("comparsion of actual and predicted price:\n",compare)

dt_mse = mean_squared_error(y_test, y_pred_dt)
print("Mean Squared Error of Decision Tree:", dt_mse)
```

```python
#control tree complexity using hyperparameter:
dt_model_1= DecisionTreeRegressor(max_depth=3,random_state=42)
dt_model_2= DecisionTreeRegressor(max_depth=5,min_samples_split=10,random_state=42)
dt_model_3= DecisionTreeRegressor(max_depth=10,min_samples_split=20,min_samples_leaf=5,random_state=42)

dt_model_1.fit(x_train,y_train)
dt_model_2.fit(x_train,y_train)
dt_model_3.fit(x_train,y_train)

pred_1= dt_model_1.predict(x_test)
pred_2= dt_model_2.predict(x_test)
pred_3= dt_model_3.predict(x_test)

mse1=mean_squared_error(y_test,pred_1)
mse2=mean_squared_error(y_test,pred_2)
mse3=mean_squared_error(y_test,pred_3)

print("mse of model 1:",mse1)
print("mse of model 2:",mse2)
print("mse of model 3:",mse3)
```

```python
#Random Forest Regression:
from sklearn.ensemble import RandomForestRegressor
rf= RandomForestRegressor(n_estimators= 100, random_state= 42)
rf.fit(x_train,y_train)
y_pred_rf= rf.predict(x_test)

compare2 = pd.DataFrame({'actual prices':y_test.head().values,
                        'predicted prices':y_pred_rf[:5]})
print("comparsion of actual and predicted price:\n",compare2)

rf_mse = mean_squared_error(y_test, y_pred_rf)
print("Mean Squared Error of Random forest:", rf_mse)
```

```python
#comparing single tree and ensemble performance:
model_comparsion = pd.DataFrame({'model':['Decsion Tree','random Forest'],
                                 'mse':[dt_mse,rf_mse]})
print("Model comparsion:\n", model_comparsion)
```

#Part f: Support vector Regression:


```python
#support vector regression:
from sklearn.svm import SVR
svr_linear = SVR(kernel ='linear')
svr_linear.fit(x_train_scaled,y_train)
y_pred_svrlinear= svr_linear.predict(x_test_scaled)

svr_rbf = SVR(kernel ='rbf')
svr_rbf.fit(x_train_scaled,y_train)
y_pred_svrrbf= svr_rbf.predict(x_test_scaled)

linear_mse= mean_squared_error(y_test,y_pred_svrlinear)
print("Linear svm=r mse:",linear_mse)
rbf_mse= mean_squared_error(y_test,y_pred_svrrbf)
print("Rbf svm=r mse:",rbf_mse)
```

```python
#Tune hyperparameters :
param_grid={'C':[0.1,1,10,100],
            'gamma':['scale','auto',0.01,0.1],
            'epsilon':[0.1,0.2,0.5]}
svr2=SVR(kernel='rbf')
svr2_grid= GridSearchCV(svr2,param_grid,cv=5,scoring='neg_mean_squared_error',n_jobs=-1)
svr2_grid.fit(x_train_scaled,y_train)
print("Best parameters:", svr2_grid.best_params_)
print("best cv mse:",-svr2_grid.best_score_)


```

```python
# compare svr performance with linear and tree based model:
#prediction of ridge and lasso
ridge_pred= best_ridge.predict(x_test_scaled)
lasso_pred= best_lasso.predict(x_test_scaled)

model_comparsion2= pd.DataFrame({'Model':['ridge','lasso','decision tree','random forest','linear svr','rbf svr'],
                                 'MSE':[mean_squared_error(y_test,ridge_pred),mean_squared_error(y_test,lasso_pred),dt_mse,rf_mse,linear_mse,rbf_mse]})
print(model_comparsion2)
```

#Part G: Model Comparison and Evaluation

```python
#evaluate all models:
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
predictions= {'Ridge':ridge_pred,
              'lasso':lasso_pred,
              'decision tree':y_pred_dt,
              'random forest':y_pred_rf,
              'linear svr':y_pred_svrlinear,
              'Rbf svr':y_pred_svrrbf}

results=[]
for name,pred in predictions.items():
  mse=mean_squared_error(y_test,pred)
  mae=mean_absolute_error(y_test,pred)
  rmse=np.sqrt(mse)
  r2=r2_score(y_test,pred)

  results.append({'Model':name,'MSe':mse,'MAE':mae,'RMSE':rmse,'R2 SCORE':r2})
  evaluation_df=pd.DataFrame(results)

print(evaluation_df.round(4))
```

Ridge and Lasso Regression showed similar performance with MSE values of approximately 6.54 × 10¹². Random Forest performed better than Decision Tree, achieving the lowest MSE of 5.84 × 10¹². Both SVR models recorded significantly higher errors. Overall, Random Forest was the best-performing model, while SVR models showed the weakest performance for this dataset.

Overfitting and underfitting:

Ridge and Lasso Regression showed good generalization with low test errors. Decision Tree may be prone to overfitting if its depth is not controlled. Random Forest reduced overfitting by combining multiple trees and achieved the best test performance. Both SVR models showed signs of underfitting because of their very high MSE values and negative R² scores.
